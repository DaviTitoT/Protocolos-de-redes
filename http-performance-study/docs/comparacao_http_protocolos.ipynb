# Comparação de protocolos HTTP e TCP em um cenário hospitalar

Este notebook apresenta uma análise parcial do projeto, com foco em arquitetura distribuída, benchmark prático e discussão sobre a escolha do protocolo mais adequado para integrações hospitalares.

## 1. Introdução
Em ambientes hospitalares, a comunicação entre sistemas é crítica. A escolha do protocolo impacta confiabilidade, interoperabilidade e facilidade de integração.

## 2. Fundamentação
HTTP é amplamente usado em integrações modernas, enquanto TCP é mais direto e de baixo nível. O HTTP oferece maior facilidade para interoperar com serviços web e APIs.

## 3. Cenário
O cenário simula um cliente enviando requisições para um módulo de integração hospitalar, medindo tempo de resposta e viabilidade prática.

## 4. Metodologia
Foram executados benchmarks locais em Python com múltiplas requisições. Os resultados foram salvos em CSV e apresentados em gráficos.

## 5. Benchmark HTTP
O benchmark HTTP utiliza o servidor e cliente implementados com Flask e requests.

## 6. Benchmark TCP
O benchmark TCP utiliza sockets para uma comparação direta com o modelo HTTP.

## 7. Gráficos e análise parcial
A análise parcial aponta que HTTP é vantajoso quando a prioridade é interoperabilidade, rastreabilidade e integração com sistemas distribuídos.
